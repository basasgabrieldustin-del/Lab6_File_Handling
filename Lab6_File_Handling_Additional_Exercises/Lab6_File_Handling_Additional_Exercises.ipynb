```python
import csv
import json
from collections import Counter
from pathlib import Path

LAST_NAME = "Basas"
STUDENT_ID = "TUPM-26-3492"

SEED_DIGIT = 2
ID_SUM = 26

RECORD_LIMIT = 1000 + ID_SUM
TARGET_CLASS = (SEED_DIGIT % 4) + 2

WORK_DIR = Path(".")
DATA_DIR = WORK_DIR / "data"
OUTPUT_DIR = WORK_DIR / "output"

dataset_path = DATA_DIR / "weblog.csv"

status_counts = Counter()
url_counts = Counter()
filtered_records = []

with open(dataset_path, "r", encoding="utf-8") as file:
    reader = csv.DictReader(file)

    print("Columns:", reader.fieldnames)

    for i, row in enumerate(reader):
        if i >= RECORD_LIMIT:
            break

        try:
            status = int(row["Staus"])
        except:
            continue

        if status // 100 == TARGET_CLASS:
            filtered_records.append(row)
            status_counts[status] += 1

            if "URL" in row:
                url_counts[row["URL"]] += 1

OUTPUT_DIR.mkdir(exist_ok=True)

output_file = OUTPUT_DIR / "filtered_logs.csv"

if filtered_records:
    with open(output_file, "w", newline="", encoding="utf-8") as out:
        writer = csv.DictWriter(
            out,
            fieldnames=filtered_records[0].keys()
        )
        writer.writeheader()
        writer.writerows(filtered_records)

print("Records found:", len(filtered_records))
print("Status counts:", dict(status_counts))
print("Top URLs:", url_counts.most_common(5))

import csv

with open("data/weblog.csv", "r", encoding="utf-8") as file:
    reader = csv.DictReader(file)

    for i, row in enumerate(reader):
        print(row)
        if i == 4:
            break
```

```python
import csv
from pathlib import Path

TARGET_OS = "Android"
TARGET_CLASS = 3

selected = []

with open("data/user_behavior_dataset.csv", "r", encoding="utf-8") as file:
    reader = csv.DictReader(file)

    for row in reader:
        if (
            row["Operating System"] == TARGET_OS
            and int(row["User Behavior Class"]) == TARGET_CLASS
        ):
            selected.append(row)

print("Records Found:", len(selected))

if selected:
    battery_total = sum(
        float(row["Battery Drain (mAh/day)"])
        for row in selected
    )

    data_total = sum(
        float(row["Data Usage (MB/day)"])
        for row in selected
    )

    print(
        "Average Battery Drain:",
        round(battery_total / len(selected), 2)
    )

    print(
        "Average Data Usage:",
        round(data_total / len(selected), 2)
    )
```

```python
import csv
from collections import Counter
from pathlib import Path

LAST_NAME = "Basas"
STUDENT_ID = "TUPM-26-3492"

SEED_DIGIT = 2
ID_SUM = 26

TARGET_YEAR = 2015 + SEED_DIGIT

WORK_DIR = Path(".")
DATA_DIR = WORK_DIR / "data"
OUTPUT_DIR = WORK_DIR / "output"

dataset_path = DATA_DIR / "Global_Cybersecurity_Threats_2015-2024.csv"

attack_counts = Counter()
records = []

with open(dataset_path, "r", encoding="utf-8") as file:
    reader = csv.DictReader(file)

    print("Columns:", reader.fieldnames)

    for row in reader:
        try:
            if int(row["Year"]) == TARGET_YEAR:
                records.append(row)

                if "Attack Type" in row:
                    attack_counts[row["Attack Type"]] += 1
        except:
            pass

print("Target Year:", TARGET_YEAR)
print("Records Found:", len(records))

if records:
    print("Attack Types:")
    print(dict(attack_counts))

    total_loss = 0

for row in records:
    try:
        total_loss += float(row["Financial Loss (in Million $)"])
    except:
        pass

average_loss = total_loss / len(records) if records else 0

print("Target Year:", TARGET_YEAR)
print("Records Found:", len(records))
print("Total Financial Loss:", round(total_loss, 2))
print("Average Financial Loss:", round(average_loss, 2))
print("Attack Types:")
print(dict(attack_counts))

top5 = sorted(
    records,
    key=lambda x: float(x["Financial Loss (in Million $)"]),
    reverse=True
)[:5]

print("\nTop 5 Costliest Incidents:")

for i, incident in enumerate(top5, 1):
    print(
        i,
        incident["Country"],
        incident["Attack Type"],
        incident["Financial Loss (in Million $)"]
    )
```
